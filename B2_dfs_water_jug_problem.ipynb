```python
cap_a = 4
cap_b = 3
target = 2


def next_states(a, b):
    pour_ab = min(a, cap_b - b)
    pour_ba = min(b, cap_a - a)
    return [
        (cap_a, b),
        (a, cap_b),
        (0, b),
        (a, 0),
        (a - pour_ab, b + pour_ab),
        (a + pour_ba, b - pour_ba),
    ]


def dfs(state, visited, path):
    if state[0] == target:
        return path
    visited.add(state)
    for nxt in next_states(*state):
        if nxt not in visited:
            result = dfs(nxt, visited, path + [nxt])
            if result:
                return result
    return None


solution = dfs((0, 0), set(), [(0, 0)])

if solution:
    print("Jug capacities:", cap_a, "and", cap_b, "| Target:", target)
    for i, state in enumerate(solution):
        print("Step", i, ": (4L jug, 3L jug) =", state)
else:
    print("No solution")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical B2 : Depth First Search - Water Jug Problem
#
# OBJECTIVE:
#   Use depth first search to solve the water jug problem with a 4 litre jug, a 3 litre jug and a target of 2 litres.
#
# THEORY:
#   DFS explores one branch as deeply as possible before backtracking.
#   A state is the pair (amount in the 4 litre jug, amount in the 3 litre jug).
#   The allowed operations are fill a jug, empty a jug and pour one jug into the other.
#   A visited set prevents the search from repeating states and looping forever.
#
# ALGORITHM:
#   1. Set the jug capacities to 4 and 3 and the target to 2.
#   2. Start from the state (0, 0).
#   3. If the first jug contains the target, return the path.
#   4. Mark the state as visited.
#   5. Generate the next states by filling, emptying and pouring the jugs.
#   6. Recurse on each unvisited state with the updated path.
#   7. Print the solution steps if a solution is found.
#
# CONCLUSION:
#   DFS finds a valid sequence that leaves 2 litres in the 4 litre jug.
#   The path is not guaranteed to be the shortest because DFS does not explore level by level.
# ===== END THEORY =====
```
