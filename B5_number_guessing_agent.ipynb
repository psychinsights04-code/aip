```python
low = 1
high = 100
tries = 0

print("Think of a number between", low, "and", high)

while low <= high:
    guess = (low + high) // 2
    tries += 1
    answer = input("Is it " + str(guess) + "? (h = higher, l = lower, c = correct): ").strip().lower()
    if answer == "c":
        print("Guessed your number", guess, "in", tries, "tries")
        break
    elif answer == "h":
        low = guess + 1
    elif answer == "l":
        high = guess - 1
    else:
        print("Please enter h, l or c")
        tries -= 1
else:
    print("Your answers were inconsistent, I could not find the number")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical B5 : Number Guessing Agent
#
# OBJECTIVE:
#   Design an intelligent agent that guesses a number using higher or lower feedback.
#
# THEORY:
#   The agent uses binary search to reduce the possible range after every guess.
#   It keeps a low bound and a high bound and always guesses the middle value.
#   Feedback of higher or lower discards half of the remaining numbers.
#   For the range 1 to 100 the agent needs at most 7 guesses.
#
# ALGORITHM:
#   1. Set low = 1, high = 100 and tries = 0.
#   2. While low is less than or equal to high, guess the middle value.
#   3. Increase the number of tries.
#   4. Ask the user whether the guess is higher, lower or correct.
#   5. If correct, stop; if higher, set low = guess + 1; if lower, set high = guess - 1.
#   6. Repeat the loop.
#   7. Print the number of tries, or report inconsistent answers.
#
# CONCLUSION:
#   Binary search guesses any number between 1 and 100 in at most 7 tries.
#   The agent is efficient because it halves the search space after every guess.
# ===== END THEORY =====
```
