```python
N = 8
solutions = []


def safe(cols, row, col):
    for r in range(row):
        c = cols[r]
        if c == col or abs(c - col) == abs(r - row):
            return False
    return True


def place(cols, row):
    if row == N:
        solutions.append(cols[:])
        return
    for col in range(N):
        if safe(cols, row, col):
            cols.append(col)
            place(cols, row + 1)
            cols.pop()


place([], 0)

print("Total solutions:", len(solutions))
print("First solution (column of queen in each row):", solutions[0])
print()
for col in solutions[0]:
    print(" ".join("Q" if c == col else "." for c in range(N)))








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical B3 : Backtracking - 8 Queens Problem
#
# OBJECTIVE:
#   Solve the 8 queens constraint satisfaction problem using backtracking.
#
# THEORY:
#   The 8 queens problem requires placing eight queens on an 8x8 board so that no two attack each other.
#   A queen attacks along its row, column and both diagonals.
#   The algorithm places one queen per row and tries every column in that row.
#   If a column is unsafe, it is skipped; if no column works, the previous queen is removed (backtracking).
#
# ALGORITHM:
#   1. Start with an empty list that stores the column of the queen in each row.
#   2. For the current row, try every column from 0 to 7.
#   3. Check that the column is not used and that no earlier queen lies on the same diagonal.
#   4. If the position is safe, place the queen and move to the next row.
#   5. When all 8 rows are filled, record the solution.
#   6. If no column works, remove the last queen and return to the previous row.
#   7. Print the total number of solutions and the first solution board.
#
# CONCLUSION:
#   Backtracking finds all 92 solutions of the 8 queens problem.
#   Rejecting unsafe positions early greatly reduces the search.
# ===== END THEORY =====
```
