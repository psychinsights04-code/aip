```python
wins = [(0, 1, 2), (3, 4, 5), (6, 7, 8), (0, 3, 6), (1, 4, 7), (2, 5, 8), (0, 4, 8), (2, 4, 6)]


def winner(b):
    for x, y, z in wins:
        if b[x] != " " and b[x] == b[y] == b[z]:
            return b[x]


def minimax(b, ai):
    w = winner(b)
    if w or " " not in b:
        return (w == "O") - (w == "X")
    scores = []
    for i in range(9):
        if b[i] == " ":
            b[i] = "O" if ai else "X"
            scores.append(minimax(b, not ai))
            b[i] = " "
    return max(scores) if ai else min(scores)


def best_move(b):
    best = None
    for i in range(9):
        if b[i] == " ":
            b[i] = "O"
            score = minimax(b, False)
            b[i] = " "
            if best is None or score > best[0]:
                best = (score, i)
    return best[1]


def show(b):
    for i in range(0, 9, 3):
        print(" " + " | ".join(b[i:i + 3]))
        if i < 6:
            print("---+---+---")


board = [" "] * 9
print("You are X, computer is O. Enter a position from 1 to 9.")
show(board)

while True:
    pos = input("Your move: ")
    if not (pos.isdigit() and 1 <= int(pos) <= 9 and board[int(pos) - 1] == " "):
        print("Invalid move, try again")
        continue
    board[int(pos) - 1] = "X"
    if winner(board) or " " not in board:
        show(board)
        break
    board[best_move(board)] = "O"
    show(board)
    if winner(board) or " " not in board:
        break

result = winner(board)
print("You win!" if result == "X" else "Computer wins!" if result == "O" else "It is a draw!")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical B7 : Minimax - Tic Tac Toe
#
# OBJECTIVE:
#   Develop a game-playing agent for Tic-Tac-Toe using the minimax algorithm.
#
# THEORY:
#   Minimax is used for two-player zero-sum games in which one player maximises and the other minimises the score.
#   The computer plays O and tries to maximise the score, while the human plays X and tries to minimise it.
#   A terminal state is scored +1 for an O win, -1 for an X win and 0 for a draw.
#   The algorithm explores all possible moves and chooses the move with the best guaranteed outcome.
#
# ALGORITHM:
#   1. Define the eight winning triples and the 3x3 board.
#   2. The winner function returns X, O or None for the board.
#   3. minimax(board, is_ai) returns the score of a terminal state; otherwise it tries every empty cell.
#   4. The AI chooses the maximum score and the human chooses the minimum score.
#   5. best_move places O in every empty cell, calls minimax and keeps the highest score.
#   6. Read and validate the human move, then let the AI play.
#   7. Repeat until a win or a draw and print the result.
#
# CONCLUSION:
#   Minimax makes the computer unbeatable because it examines the full game tree.
#   The same idea can be made faster by adding alpha-beta pruning.
# ===== END THEORY =====
```
