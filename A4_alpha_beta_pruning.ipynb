```python
values = [3, 5, 6, 9, 1, 2, 0, -1]
depth = 3
visited = []


def alphabeta(d, index, is_max, alpha, beta):
    if d == depth:
        visited.append(values[index])
        return values[index]
    if is_max:
        best = float("-inf")
        for i in range(2):
            val = alphabeta(d + 1, index * 2 + i, False, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    best = float("inf")
    for i in range(2):
        val = alphabeta(d + 1, index * 2 + i, True, alpha, beta)
        best = min(best, val)
        beta = min(beta, best)
        if beta <= alpha:
            break
    return best


result = alphabeta(0, 0, True, float("-inf"), float("inf"))
print("Leaf values:", values)
print("Optimal value at root:", result)
print("Leaves visited:", visited)
print("Leaves pruned:", len(values) - len(visited))








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A4 : Adversarial Search - Alpha-Beta Pruning
#
# OBJECTIVE:
#   Implement adversarial search with alpha-beta pruning and observe the leaves that are pruned.
#
# THEORY:
#   Minimax is used for two-player zero-sum games where MAX and MIN take alternate turns.
#   Alpha is the best value found so far for MAX and beta is the best value found so far for MIN.
#   A branch is pruned when beta becomes less than or equal to alpha, because it cannot affect the result.
#   Alpha-beta pruning returns the same value as minimax while visiting fewer nodes.
#
# ALGORITHM:
#   1. Take the leaf utility values and build a binary game tree of depth 3.
#   2. Start at the root as a MAX node with alpha = -infinity and beta = +infinity.
#   3. At a MAX node, evaluate children and update alpha; prune when beta <= alpha.
#   4. At a MIN node, evaluate children and update beta; prune when beta <= alpha.
#   5. Continue recursively until the leaf values are reached.
#   6. Return the root value and record the visited and pruned leaves.
#
# CONCLUSION:
#   The root value is 5, which is the same as plain minimax.
#   Three leaves are pruned, so alpha-beta pruning saves unnecessary computation.
# ===== END THEORY =====
```
