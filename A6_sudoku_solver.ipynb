```python
puzzle = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9],
]


def options(b, r, c):
    used = set(b[r]) | {b[i][c] for i in range(9)}
    br, bc = r - r % 3, c - c % 3
    used |= {b[br + i][bc + j] for i in range(3) for j in range(3)}
    return [v for v in range(1, 10) if v not in used]


def propagate(b):
    changed = True
    while changed:
        changed = False
        for r in range(9):
            for c in range(9):
                if b[r][c] == 0:
                    o = options(b, r, c)
                    if not o:
                        return False
                    if len(o) == 1:
                        b[r][c], changed = o[0], True
    return True


def solve(board):
    b = [row[:] for row in board]
    if not propagate(b):
        return None
    empty = [(options(b, r, c), r, c) for r in range(9) for c in range(9) if b[r][c] == 0]
    if not empty:
        return b
    opts, r, c = min(empty, key=lambda x: len(x[0]))
    for v in opts:
        b[r][c] = v
        result = solve(b)
        if result:
            return result
    return None


def show(b):
    for i, row in enumerate(b):
        if i and i % 3 == 0:
            print("------+-------+------")
        print(" | ".join(" ".join(map(str, row[j:j + 3])) for j in (0, 3, 6)))


print("Puzzle:")
show(puzzle)
solution = solve(puzzle)
print("\nSolution:" if solution else "\nNo solution")
if solution:
    show(solution)








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A6 : Sudoku Solver - Backtracking and Constraint Propagation
#
# OBJECTIVE:
#   Solve a 9x9 Sudoku puzzle using backtracking combined with constraint propagation.
#
# THEORY:
#   Sudoku is a constraint satisfaction problem in which every row, column and 3x3 box must contain 1 to 9 once.
#   For each empty cell, the allowed values are those not present in its row, column or box.
#   Constraint propagation fills a cell immediately when it has only one allowed value.
#   Backtracking tries candidates for the most constrained cell and undoes a choice if it leads to a dead end.
#
# ALGORITHM:
#   1. Read the puzzle, using 0 for empty cells.
#   2. For every empty cell, compute the allowed values from its row, column and box.
#   3. Repeatedly fill cells that have exactly one allowed value (propagation).
#   4. If any empty cell has no allowed value, report a contradiction and backtrack.
#   5. Choose the empty cell with the fewest allowed values.
#   6. Try each allowed value, solve recursively, and undo the value on failure.
#   7. Print the completed grid.
#
# CONCLUSION:
#   Constraint propagation quickly fixes forced cells and reduces the search space.
#   Backtracking completes the puzzle correctly and efficiently.
# ===== END THEORY =====
```
