```python
env = {
    "A" : "dirty",
    "B" : "dirty"
}

location = "A"

print(f"initial env : {env}")
print(f"agent start posi : {location}")

def vac_agent(location,status):
    if status == "dirty" :
        return "suck"
    elif location == "A":
        return "move right"
    else :
        return "move left"


#while env.status() == dirty :
while 'dirty' in env.values() :
    status = env[location]
    action = vac_agent(location,status)
    print(f"{location}-{status}-{action}")
    if action == "suck":
        env[location] = "clean"
    elif action == "move right":
        location = "B"
    else :
        location = "A"

print(f"the final status {env}")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A1 : Simple Reflex Agent - Vacuum Cleaner World
#
# OBJECTIVE:
#   Design and implement a simple reflex agent for the vacuum cleaner world and analyse its rational behaviour.
#
# THEORY:
#   A simple reflex agent selects an action using only the current percept and a set of condition-action rules.
#   The vacuum world has two rooms A and B, and each room is either dirty or clean.
#   The agent sucks when the current room is dirty; otherwise it moves to the other room.
#   It has no memory, so it works well only in a fully observable, deterministic environment.
#
# ALGORITHM:
#   1. Initialise the environment with rooms A and B and place the agent in room A.
#   2. Read the status of the current room.
#   3. If the room is dirty, perform Suck and mark it clean.
#   4. Otherwise, if the agent is in A, move right to B; else move left to A.
#   5. Repeat steps 2-4 until no room is dirty.
#   6. Display the final environment.
#
# CONCLUSION:
#   The agent cleans both rooms using only the current percept.
#   It is simple and fast, but cannot handle large or partially observable environments without memory.
# ===== END THEORY =====
```
