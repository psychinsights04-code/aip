```python
import re

rules = [
    (r"\b(hi|hello|hey)\b", "Hello! How can I help you today?"),
    (r"my name is (\w+)", "Nice to meet you, {0}!"),
    (r"how are you", "I am doing fine, thanks for asking."),
    (r"your name", "I am a simple rule based chatbot."),
    (r"\b(admission|admissions)\b", "Admissions start in June. Please visit the college office."),
    (r"\b(fee|fees)\b", "You can get the fee structure from the accounts section."),
    (r"\b(exam|exams)\b", "The exam timetable is available on the notice board."),
    (r"i feel (\w+)", "Why do you feel {0}?"),
    (r"\b(thanks|thank you)\b", "You are welcome!"),
    (r"\b(bye|exit|quit)\b", "Goodbye! Have a nice day."),
]


def reply(text):
    for pattern, answer in rules:
        match = re.search(pattern, text, re.IGNORECASE)
        if match:
            return answer.format(*match.groups())
    return "Sorry, I did not understand that. Can you rephrase?"


print("Chatbot started. Type bye to exit.")
while True:
    user = input("You: ").strip()
    print("Bot:", reply(user))
    if re.search(r"\b(bye|exit|quit)\b", user, re.IGNORECASE):
        break








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical B4 : Rule-Based Chatbot
#
# OBJECTIVE:
#   Develop a simple chatbot using predefined rules and pattern matching.
#
# THEORY:
#   A rule-based chatbot stores a list of patterns and the response to send when a pattern matches.
#   Regular expressions are used to match keywords and to capture parts of the user input.
#   Captured groups are substituted into the response to make it feel personalised.
#   If no rule matches, the chatbot returns a default message.
#
# ALGORITHM:
#   1. Define a list of rules, where each rule contains a regular expression and a response.
#   2. Read a line of input from the user.
#   3. Search each rule in the input using case-insensitive matching.
#   4. If a rule matches, format the response with the captured groups and return it.
#   5. If no rule matches, return a default message.
#   6. Repeat until the user types bye, exit or quit.
#
# CONCLUSION:
#   Pattern matching gives fast responses for known questions.
#   The chatbot has no learning ability and is limited to the rules written by the programmer.
# ===== END THEORY =====
```
