```python
grid = [
    ["S", ".", ".", "."],
    [".", ".", "#", "."],
    ["#", ".", "#", "."],
    [".", "#", "#", "G"],
]

rows = len(grid)
columns = len(grid[0])
start = (0, 0)
path = []

# internal state / model of the agent: cells it has already visited
memory = set()

moves = [(1, 0), (0, 1), (-1, 0), (0, -1)]


def agent(position):
    row, column = position
    memory.add(position)          # update internal model
    path.append(position)

    if grid[row][column] == "G":
        return True

    for dr, dc in moves:
        nr, nc = row + dr, column + dc
        if (0 <= nr < rows and 0 <= nc < columns
                and grid[nr][nc] != "#"
                and (nr, nc) not in memory):
            if agent((nr, nc)):
                return True

    path.pop()                    # dead end: backtrack
    return False


if agent(start):
    print("Goal found")
    print("Path:", path)
    print("Steps:", len(path) - 1)
else:
    print("No path found")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A2 : Model-Based Agent - Grid Navigation
#
# OBJECTIVE:
#   Implement a model-based agent that navigates a grid containing obstacles using internal state representation.
#
# THEORY:
#   A model-based agent keeps an internal model of the world and updates it as it acts.
#   Here the grid has a start S, a goal G, free cells and blocked cells marked #.
#   The agent stores every visited cell in memory, which prevents it from visiting the same cell again.
#   Depth-first search with backtracking is used to explore the grid until the goal is reached.
#
# ALGORITHM:
#   1. Define the grid with start, goal and obstacle cells.
#   2. Initialise the agent memory (visited set) and the path list.
#   3. Mark the current cell visited and add it to the path.
#   4. If the current cell is the goal, return the path.
#   5. Try each neighbour (down, right, up, left) that is inside the grid, free and not in memory.
#   6. Recursively move to a valid neighbour; if it leads to the goal, return true.
#   7. If no neighbour works, remove the current cell from the path (backtrack).
#
# CONCLUSION:
#   The internal state prevents repeated visits and enables backtracking out of dead ends.
#   A valid path is found, although it is not guaranteed to be the shortest path.
# ===== END THEORY =====
```
