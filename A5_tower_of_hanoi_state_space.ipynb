```python

from collections import deque

n = 3
start = (0, 0, 0)
goal = (2, 2, 2)
pegs = "ABC"

q = deque([(start, [])])
visited = {start}
path = []

while q:
    state, path = q.popleft()

    if state == goal:
        break

    for i in range(n):
        src = state[i]

        if any(state[j] == src for j in range(i)):
            continue

        for dst in range(3):
            if dst == src:
                continue

            if any(state[j] == dst for j in range(i)):
                continue

            new = list(state)
            new[i] = dst
            new = tuple(new)

            if new not in visited:
                visited.add(new)
                q.append((new, path + [(i+1, src, dst)]))

for step, (disk, src, dst) in enumerate(path, 1):
    print("Move", step, ": Disk", disk, "from", pegs[src], "to", pegs[dst])

print("Total moves:", len(path))








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A5 : Tower of Hanoi - State Space Search
#
# OBJECTIVE:
#   Solve the Tower of Hanoi problem using state space search and find the minimum number of moves.
#
# THEORY:
#   In state space search, each state describes the position of every disk on the three pegs.
#   A legal move takes the top disk of one peg and places it on another peg, never on a smaller disk.
#   Breadth-first search explores states level by level, so the first goal state found is optimal.
#   For n disks the minimum number of moves is 2^n - 1.
#
# ALGORITHM:
#   1. Represent a state as a tuple giving the peg of each disk.
#   2. Set the start state with all disks on peg A and the goal state with all disks on peg C.
#   3. Insert the start state into a queue with an empty move list and mark it visited.
#   4. Remove the first state from the queue.
#   5. If it is the goal state, stop.
#   6. For every top disk, try moving it to each other peg where the move is legal.
#   7. Add each new state to the queue with the updated move list.
#   8. Print the moves and the total number of moves.
#
# CONCLUSION:
#   BFS finds the optimal solution of 7 moves for 3 disks.
#   The experiment demonstrates state space search and optimal planning.
# ===== END THEORY =====
```
