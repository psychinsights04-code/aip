```python
import random
import math

random.seed(3)
n = 8
k = 3
cities = [(random.randint(0, 100), random.randint(0, 100)) for _ in range(n)]


def tour_length(tour):
    total = 0
    for i in range(n):
        total += math.dist(cities[tour[i]], cities[tour[(i + 1) % n]])
    return total


def neighbours(tour):
    result = []
    for i in range(n - 1):
        for j in range(i + 1, n):
            t = list(tour)
            t[i], t[j] = t[j], t[i]
            result.append(tuple(t))
    return result


beam = []
for _ in range(k):
    t = list(range(n))
    random.shuffle(t)
    beam.append(tuple(t))

best = min(beam, key=tour_length)
print("Cities:", cities)
print("Initial best length:", round(tour_length(best), 2))

step = 0
while True:
    pool = set()
    for t in beam:
        pool.update(neighbours(t))
    new_beam = sorted(pool, key=tour_length)[:k]
    if tour_length(new_beam[0]) >= tour_length(best):
        break
    beam = new_beam
    best = new_beam[0]
    step += 1
    print("Step", step, "best length:", round(tour_length(best), 2))

print()
print("Best tour:", list(best) + [best[0]])
print("Tour length:", round(tour_length(best), 2))








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A7 : Local Beam Search - Travelling Salesman Problem
#
# OBJECTIVE:
#   Solve the Travelling Salesman Problem using local beam search.
#
# THEORY:
#   Local beam search is a local search technique that keeps the k best states instead of only one.
#   Each iteration expands all k states and keeps the k best neighbours for the next iteration.
#   In the TSP, a tour must visit every city once and return to the starting city with minimum length.
#   Neighbours of a tour are generated by swapping two cities in the tour.
#
# ALGORITHM:
#   1. Generate n random cities and choose the beam width k.
#   2. Create k random tours as the initial beam.
#   3. Compute the length of each tour.
#   4. Generate all neighbours of every tour by swapping two cities.
#   5. Keep the k shortest tours as the new beam.
#   6. Stop when the best tour length does not improve.
#   7. Print the best tour and its length.
#
# CONCLUSION:
#   Local beam search improves the tour quickly and gives a near-optimal solution.
#   The final result may depend on the initial random tours and the beam width.
# ===== END THEORY =====
```
