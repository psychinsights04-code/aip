```python
import heapq

goal = (1, 2, 3, 8, 0, 4, 7, 6, 5)


def heuristic(state):
    total = 0
    for i, v in enumerate(state):
        if v != 0:
            g = goal.index(v)
            total += abs(i // 3 - g // 3) + abs(i % 3 - g % 3)
    return total


def neighbors(state):
    i = state.index(0)
    r, c = divmod(i, 3)
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state)
            s[i], s[j] = s[j], s[i]
            result.append(tuple(s))
    return result


def astar(start):
    heap = [(heuristic(start), 0, start, [start])]
    best_g = {}
    while heap:
        f, g, state, path = heapq.heappop(heap)
        if state == goal:
            return path
        if state in best_g and best_g[state] <= g:
            continue
        best_g[state] = g
        for n in neighbors(state):
            heapq.heappush(heap, (g + 1 + heuristic(n), g + 1, n, path + [n]))
    return None


def show(state):
    for i in range(0, 9, 3):
        print(" ".join(str(x) if x != 0 else "_" for x in state[i:i + 3]))
    print()


start = (2, 8, 3, 1, 6, 4, 7, 0, 5)
path = astar(start)

if path:
    print("Solved in", len(path) - 1, "moves\n")
    for step, state in enumerate(path):
        print("Step", step)
        show(state)
else:
    print("No solution")








































# ===== THEORY (copy to journal, then delete this block) =====
# Practical A3 : A* Search - 8 Puzzle
#
# OBJECTIVE:
#   Implement A* search to solve the 8-puzzle problem using the Manhattan distance heuristic.
#
# THEORY:
#   A* is an informed search algorithm that expands the node with the lowest f(n) = g(n) + h(n).
#   g(n) is the cost from the start and h(n) is the estimated cost to the goal.
#   The Manhattan distance is admissible because it never overestimates the real cost.
#   A min-heap (priority queue) stores the states in increasing order of f(n).
#
# ALGORITHM:
#   1. Represent the puzzle as a tuple of 9 tiles, where 0 is the blank tile.
#   2. Define the goal state and the Manhattan distance heuristic.
#   3. Push the start state into a min-heap with f = h(start).
#   4. Pop the state with the smallest f value.
#   5. If it is the goal state, return the stored path.
#   6. Generate neighbours by moving the blank up, down, left or right.
#   7. Push each neighbour with g+1 and f = g+1+h, skipping states with a better g already.
#   8. Display the moves if a solution is found.
#
# CONCLUSION:
#   A* finds the optimal 5-move solution for the given puzzle.
#   The Manhattan heuristic reduces the number of states explored.
# ===== END THEORY =====
```
